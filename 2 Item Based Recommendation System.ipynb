# Python codes for recomendation system to determine ratings based on movie id (item data)

```python
# Import Python libraries
import pandas as pd
import numpy as np
```

```python
# Load Dataset
Col_names = ['User_id', 'Movie_id', 'Ratings', 'Unique_timestamp']
Ratings = pd.read_csv('ml-100k/u.data', sep = '\t', names = Col_names, encoding = 'latin_1')
Ratings
```

```python
N_Items = Ratings.Movie_id.unique().shape[0]
print(f"There are {N_Items} in the dataset")
```

```python
# Create pivot table based on user id, movie id and rating
Pivot_table = Ratings.pivot_table (index = 'User_id', columns = 'Movie_id', values = 'Ratings')
Pivot_table
```

```python
# Replace Nan with 0
Data_matrix = Pivot_table.replace(np.nan, 0)
Data_matrix
```

```python
# Find cosine similarity for users
from sklearn.metrics.pairwise import pairwise_distances
Item_similarity = pairwise_distances(Data_matrix.T, metric = 'cosine')
Item_similarity
```

```python
# Score calculation
def prediction(Ratings, Similarity, type = 'item'):
    Item_ratings_mean = Ratings.mean(axis = 1).to_numpy()
    Ratings_diff = Ratings - Item_ratings_mean [:, np.newaxis]
    Predict = Ratings.dot(Similarity) / np.array([np.abs(Similarity).sum(axis=1)])
    return Predict
```

```python
# Prediction table
Item_prediction = prediction(Data_matrix, Item_similarity, type = 'item')
Item_prediction
```

```python
# Selection of input item
Input_items = 34
```

```python
# Convert user sim table into dataframe
Item_Sim_Table = pd.DataFrame(Item_similarity)
Item_Sim_Table
```

```python
# Finding similarity for user 34 using cosine table
Sim_Input_Item = Item_Sim_Table[Input_items].sort_values(ascending = True).head(5).index
Sim_Input_Item
```

```python
# Convert the similarity values into a list
Sim_Input_Item = list(Sim_Input_Item)
Sim_Input_Item
```

```python
# Select movie id from ratings table using Similarity_Input_Item
Sim_Movie_UserID_List = []
for Sim_Item in Sim_Input_Item:
    Sim = list(Ratings[Ratings['Movie_id']==Sim_Item]['User_id'])
    Sim_Movie_UserID_List.append(Sim)
```
