# Python codes for recomendation system to determine ratings based on movie id (item data)

```python
# Import Python libraries
```

```python
import pandas as pd
import numpy as np
```

```python
# Load Dataset
```

```python
Col_names = ['User_id', 'Movie_id', 'Ratings', 'Unique_timestamp']
Ratings = pd.read_csv('ml-100k/u.data', sep = '\t', names = Col_names, encoding = 'latin-1')
Ratings
```

```python
N_Items = Ratings.Movie_id.unique().shape[0]
print(f"There are {N_Items} in the dataset")
```

```python
# Create pivot table based on user id, movie id and rating
```

```python
Pivot_table = Ratings.pivot_table (index = 'User_id', columns = 'Movie_id', values = 'Ratings')
Pivot_table
```

```python
# Replace Nan with 0
```

```python
Data_matrix = Pivot_table.replace(np.nan, 0)
Data_matrix
```

```python
# Find cosine similarity for users
```

```python
from sklearn.metrics.pairwise import pairwise_distances
Item_similarity = pairwise_distances(Data_matrix, metric = 'cosine')
Item_similarity
```

```python
# Score calculation
```

```python
def prediction(Ratings, Similarity, type = 'item'):
    Item_ratings_mean = Ratings.mean(axis = 1).to_numpy()
    Ratings_diff = Ratings - Item_ratings_mean [:, np.newaxis]
    Predict = Item_ratings_mean [:, np.newaxis] + Ratings_diff.dot(Similarity) / np.abs(Ratings_diff).sum(axis=1)[:, np.newaxis]
    return Predict
```

```python
# Prediction table
```

```python
Item_prediction = prediction(Data_matrix, Item_similarity, type = 'item')
Item_prediction
```
