# Python codes for recomendation system to determine ratings - Based on user id

```python
# Import Python libraries
import pandas as pd
import numpy as np
```

```python
# Load Dataset
Col_names = ['User_id', 'Movie_id', 'Ratings', 'Unique_timestamp']
Ratings = pd.read_csv('ml-100k/u.data', sep = '\t', names = Col_names, encoding = 'latin_1')
Ratings
```

```python
# Indentifying the number of users in the data set
N_Users = Ratings.User_id.unique().shape[0]
print(f"There are {N_Users} in the dataset")
```

```python
# Create pivot table for the users in the data set based on the ratings
Pivot_table = Ratings.pivot_table (index = 'User_id', columns = 'Movie_id', values = 'Ratings')
Pivot_table
```

```python
# Replace Nan in the pivot table with 0
Data_matrix = Pivot_table.replace(np.nan, 0)
Data_matrix
```

```python
# Find cosine similarity for users
from sklearn.metrics.pairwise import pairwise_distances
User_similarity = pairwise_distances(Data_matrix, metric = 'cosine')
User_similarity
```

```python
# Score calculation
def prediction (Ratings, Similarity, type = 'user'):
    User_ratings_mean = Ratings.mean(axis = 1).to_numpy()
    Ratings_diff = Ratings - User_ratings_mean [:, np.newaxis]
    Predict = User_ratings_mean [:, np.newaxis] + Similarity.dot(Ratings_diff) / np.array([np.abs(Similarity).sum(axis=1)]).T        
    return Predict
```

```python
# Create prediction table
User_prediction = prediction(Data_matrix, User_similarity, type = 'user')
User_prediction
```

# Creating a function to find similarity among users

```python
def UserBased (Input_user, User_similarity, User_prediction, Similar_user_count, thres):
    
    # Convert user sim table to the dataframe
    User_Sim_Table = pd.DataFrame(User_similarity)

    # Finding similarity for user 34 using cosine table
    Sim_Input_User = User_Sim_Table[Input_user].sort_values(ascending = True).head(Similar_user_count).index

    # Convert the similarity values into a list
    Sim_Input_User = list(Sim_Input_User)

    # Select movie id from ratings table for the Similarity_Input_User
    Sim_User_MovieID_List = []
    for Sim_User in Sim_Input_User:
        Sim = list(Ratings[Ratings['User_id']==Sim_User]['Movie_id'])
        Sim_User_MovieID_List.append(Sim)

    # Convert all the lists as a single list
    import itertools
    Sim_User_MovieID_Single_List = list(itertools.chain.from_iterable(Sim_User_MovieID_List))

    # Find unique movie IDs from the list
    Unique_MovieID_Similar_User = set(Sim_User_MovieID_Single_List)

    # Compare the nput movie list with users watched 
    Input_User_Watched_MovieID = list(Ratings[Ratings['User_id']==Input_user]['Movie_id'].values)

    # Create a list to show movie ids to users
    Recom = []
    for per_id in Unique_MovieID_Similar_User:
        if (per_id in Input_User_Watched_MovieID):
            pass
    else:
        Recom.append(per_id)

    # Selection of the highest rate movies that would be selected by the user, based on user prediction
    Highest_rated = []
    User_pred = pd.DataFrame(User_prediction)
    User_pred_Trans = User_pred.T
    Input_User_Pre = pd.DataFrame(User_pred_Trans[Input_user])
    Input_User_Pred = Input_User_Pre.T
    for Re in Recom:
        Value = Input_User_Pred[Re].values
        if(Value >= thres):
            Highest_rated.append(Re)

    # Securing names
    I_Cols = ['Movie id', 'Movie title', 'Release Date', 'Video Release Date', 'IMDb URL', 'unknown', 'Action', 'Adventure', 'Animation', 'Childrens', 'Comedy', 'Crime', 'Documentary', 'Drama', 'Fantasy', 'Film-Noir', 'Horro', 'Musical', 'Mystery', 'Romance', 'Sci-Fi', 'Thriller', 'War', 'Western']
    Items = pd.read_csv('ml-100k/u.item', sep = '|', names = I_Cols, encoding = 'latin_1')

    # Creating movie list based on recommended movie list
    Movie_title = []
    for MovieID in Highest_rated:
        Movie = Items[Items['Movie id'] == MovieID]['Movie title'].values
        Movie_title.append(Movie)

    # Converting into a pure list
    Movie_Title_List = []
    for M in Movie_title:
        print(M)
        MV = list(M)
        Movie_Title_List.append(MV)

    # Converting into a whole list
    import itertools
    Final_Recommended_Movie = list(itertools.chain.from_iterable(Movie_Title_List))
    print ("The common Movie ID in Recom & User:", list(set(Recom) & set(Input_User_Watched_MovieID)))
    return Final_Recommended_Movie
```
