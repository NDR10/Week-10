# Python codes for recomendation system to determine ratings based on cotent

```python
# Import Python libraries
import pandas as pd
import numpy as np
```

```python
# Load Dataset
Col_names = ['movie id','movie title','release date','video release date','IMDb URL','Unknown','Action','Adventure','Animation','Children\'s',' Comedy','Crime','Documentary','Drama','Fantasy','Film-Noir','Horror','Musical','Mystery','Romance','Sci-Fi','Thriller','War','Western']
Items = pd.read_csv('ml-100k/u.item', sep = '|', names = Col_names, encoding = 'latin-1')
Items.head(4)
```

```python
# Select only the essential items
Movie_content = Items[['movie id','movie title','Action','Adventure','Animation','Children\'s',' Comedy','Crime','Documentary','Drama','Fantasy','Film-Noir','Horror','Musical','Mystery','Romance','Sci-Fi','Thriller','War','Western']]
Movie_content
```

```python
# Read the eratings dataset
R_cols = ['user id','item id','rating','timestamp']
Ratings = pd.read_csv('ml-100k/u.data', sep = '\t', names = R_cols, encoding = 'latin-1')
Ratings
```

```python
# Checking wether it is correctly receiving the data from the user id dataset
Ratings[Ratings['user id']==196][['item id','rating']]
```

```python
# Create a Pivot Table that will act as the user vector
Data_Matrix = Ratings.pivot_table(index = 'user id', columns = 'item id', values = 'rating')
Data_Matrix 
```

```python
# Replace Nan with Zero
Data_Matrix = Data_Matrix.replace(np.nan,0)
Data_Matrix 
```

```python
# Create user vector
User_Vector = Data_Matrix.T
User_Vector
```

```python
# Create movie vector
Movie_Vector = Items[['Action','Adventure','Animation','Children\'s',' Comedy','Crime','Documentary','Drama','Fantasy','Film-Noir','Horror','Musical','Mystery','Romance','Sci-Fi','Thriller','War','Western']]
```

```python
# Renaming the index
Movie_Vector.index = range(1,1683)
```

```python
# Transposing the movie vector
Movie_Vector = Movie_Vector.T
Movie_Vector
```

# Creation of content vector

```python
# Combining user vector and movie vector
```

```python
# Check the content for a random user
U1 = Ratings[Ratings['user id']==1][['item id','rating']]
U1
```

```python
# Create content for user based on movie id
U1_list = []
for i in U1['item id']:
    F = Movie_Vector[i].values
    U1_list.append(F)
```

```python
# Cnverting the dataframe
U1a = pd.DataFrame(U1_list, columns = range(1,19))
U1a
```

```python
# Creating a single vector for an user (summing each genere column)
Single_Vector = []
for j in range(1,19):    
    g = sum(U1a[j]/272)
    Single_Vector.append(g)
Single_Vector
```

```python
# Create a vector list for all users with movie vector (Created as a function)
def User_Vector(Ratings,Movie_Vector):
    Vector = []
    for user in range(1,944):
        U1 = Ratings[Ratings['user id']==user][['item id','rating']]
        U1_list = []
        for i in U1['item id']:
            F = Movie_Vector[i].values
            U1_list.append(F)
        U1a = pd.DataFrame(U1_list, columns = range(1,19))
        Single_Vector = []
        for j in range(1,19):    
            g = sum(U1a[j]/len(U1))
            Single_Vector.append(g)
        Vector.append(Single_Vector)
    return Vector
```

```python
UV = User_Vector(Ratings,Movie_Vector)
# Converting into Data sheet
UV = pd.DataFrame(UV)
UV
```

```python
# Checking the shape of user vector
UV.shape
```

```python
# Checking the shape of movie vector
Movie_Vector.shape
```

```python
# Create content vector (dot matrix) for user vector and movie vector
Content_Vector = np.dot(UV,Movie_Vector)
```

```python
# Create data frame for the content vector
Content_Vector = pd.DataFrame(Content_Vector)
```

```python
# Convert tthe data fram to allow access to the users
Content_Vector = Content_Vector.T
Content_Vector 
```

```python
# Create a data file for back up
Content_Vector.to_csv("Content_Vector.csv", index = False)
```
